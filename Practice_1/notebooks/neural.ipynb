```python
import torch
import torch.nn as nn

# Define Neural Network
class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()

        # Flatten layer: Convert 28x28 image to a 784-dimensional vector
        self.flatten = nn.Flatten()

        # Multilayer Perceptron (MLP)
        self.network = nn.Sequential(
            nn.Linear(28 * 28, 128),
            nn.ReLU(),

            nn.Linear(128, 64),
            nn.ReLU(),

            nn.Linear(64, 10)
        )

    def forward(self, x):
        x = self.flatten(x)
        x = self.network(x)
        return x
```

```python
# Create model
model = NeuralNetwork()
```

```python
# Display model architecture
print(model)
```
