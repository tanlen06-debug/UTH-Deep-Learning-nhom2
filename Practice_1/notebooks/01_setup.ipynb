## CPU and GPU

PyTorch can run on both CPU and GPU. A GPU usually makes model training faster because it can handle many calculations at the same time.

If CUDA is not available, the lab will run on the CPU.

```python
# Cell 3: Device Configuration

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Selected device:", device)

if torch.cuda.is_available():
    print("GPU name:", torch.cuda.get_device_name(0))
else:
    print("GPU is not available. Using CPU.")
```

## Tensor

A tensor is the main data structure used in PyTorch. It is similar to a NumPy array and can store vectors, matrices, images, and other numerical data.

In this example, two small tensors are created to test some basic operations.

```python
# Cell 4: Tensor Demo

x = torch.tensor([
    [1.0, 2.0],
    [3.0, 4.0]
])

y = torch.ones((2, 2))

print("Tensor x:\n", x)
print("\nTensor y:\n", y)

print("\nShape:", x.shape)
print("Data type:", x.dtype)

print("\nAddition:\n", x + y)
print("\nElement-wise multiplication:\n", x * y)
print("\nMatrix multiplication:\n", x @ y)

x = x.to(device)
print("\nTensor device:", x.device)
```
