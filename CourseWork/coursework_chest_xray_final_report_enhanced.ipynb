## 3. Preprocessing and data augmentation — Member 05

Training-time augmentation uses random horizontal flipping and rotation. Validation and test images receive deterministic resizing and normalization only.

The purpose of augmentation is to reduce overfitting by exposing the model to small appearance variations. Rotation should remain mild because large rotations can create anatomically unrealistic X-rays. The current 10° limit is acceptable for an educational experiment, though a controlled no-augmentation baseline is required to quantify its effect.


```python
IMAGE_SIZE = 224
BATCH_SIZE = 32

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

val_test_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
train_dir = DATA_ROOT / "train"
val_dir = DATA_ROOT / "val"
test_dir = DATA_ROOT / "test"

train_dataset = datasets.ImageFolder(
    root=train_dir,
    transform=train_transform
)

val_dataset = datasets.ImageFolder(
    root=val_dir,
    transform=val_test_transform
)

test_dataset = datasets.ImageFolder(
    root=test_dir,
    transform=val_test_transform
)

print("Classes:", train_dataset.classes)
print("Class mapping:", train_dataset.class_to_idx)

print("\nSố ảnh train:", len(train_dataset))
print("Số ảnh validation:", len(val_dataset))
print("Số ảnh test:", len(test_dataset))
```

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

print("Đã tạo DataLoader thành công.")
```

## 4. CNN architecture

The model uses three convolution/max-pooling blocks followed by a fully connected classifier. The executed model contains 25,784,130 trainable parameters, most of which come from flattening the 128×28×28 feature map.

This architecture is valid for the classification experiment, but it does **not** meet an age-regression rubric that explicitly requires Global Average Pooling and `Linear(1)`. For classification, a GAP-based head would also be a useful improvement because it would reduce parameters and overfitting risk.


```python
class ChestXRayCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            # Block 1
            nn.Conv2d(
                in_channels=3,
                out_channels=32,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),

            # Block 2
            nn.Conv2d(
                in_channels=32,
                out_channels=64,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),

            # Block 3
            nn.Conv2d(
                in_channels=64,
                out_channels=128,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2)
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),

            nn.Linear(
                128 * 28 * 28,
                256
            ),
            nn.ReLU(),
            nn.Dropout(p=0.5),

            nn.Linear(256, 2)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


model = ChestXRayCNN().to(device)

print(model)
```

```python
total_params = sum(
    parameter.numel()
    for parameter in model.parameters()
)

trainable_params = sum(
    parameter.numel()
    for parameter in model.parameters()
    if parameter.requires_grad
)

print("Tổng số tham số:", total_params)
print("Số tham số có thể huấn luyện:", trainable_params)
```

## 5. Training protocol

- Optimizer: Adam.
- Learning rate: 0.001 in the executed training cell.
- Epochs: 10.
- Objective: cross-entropy loss.
- Checkpoint selection: smallest validation loss.
- Reproducibility seed: 42.

The learning rate differs from the earlier configuration cell (`1e-4`). The executed training cell uses `1e-3`; this is the value that should be reported for the presented results.


```python
criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)

EPOCHS = 10
```

```python
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)

        predictions = outputs.argmax(dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

    epoch_loss = running_loss / total
    epoch_accuracy = correct / total

    return epoch_loss, epoch_accuracy
```

```python
def evaluate_model(model, loader, criterion, device):
    model.eval()

    running_loss = 0.0
    correct = 0
    total = 0

    all_labels = []
    all_predictions = []

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)

            predictions = outputs.argmax(dim=1)

            correct += (predictions == labels).sum().item()
            total += labels.size(0)

            all_labels.extend(labels.cpu().numpy())
            all_predictions.extend(predictions.cpu().numpy())

    epoch_loss = running_loss / total
    epoch_accuracy = correct / total

    return (
        epoch_loss,
        epoch_accuracy,
        all_labels,
        all_predictions
    )
```

```python
from copy import deepcopy

history = {
    "train_loss": [],
    "train_accuracy": [],
    "val_loss": [],
    "val_accuracy": []
}

best_val_loss = float("inf")
best_model_state = None

for epoch in range(EPOCHS):
    train_loss, train_accuracy = train_one_epoch(
        model,
        train_loader,
        criterion,
        optimizer,
        device
    )

    val_loss, val_accuracy, _, _ = evaluate_model(
        model,
        val_loader,
        criterion,
        device
    )

    history["train_loss"].append(train_loss)
    history["train_accuracy"].append(train_accuracy)
    history["val_loss"].append(val_loss)
    history["val_accuracy"].append(val_accuracy)

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_model_state = deepcopy(model.state_dict())

    print(
        f"Epoch [{epoch + 1}/{EPOCHS}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f}"
    )
```

## 7. Extended evaluation and error analysis — Member 06

This section distinguishes **false alarms** from **missed disease cases**, rather than relying on accuracy alone. `PNEUMONIA` is treated as the positive class:

- **True positive (TP = 384):** pneumonia correctly detected.
- **True negative (TN = 143):** normal image correctly rejected.
- **False positive (FP = 91):** a normal image is incorrectly flagged as pneumonia — *false alarm / bắt nhầm*.
- **False negative (FN = 6):** a pneumonia image is incorrectly predicted as normal — *missed case / bỏ sót*.

The following outputs are calculated from the executed test-set confusion matrix, not from invented or synthetic predictions.

```python
# Compute diagnostic metrics directly from the test predictions
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(test_labels, test_predictions, labels=[0, 1])
tn, fp, fn, tp = cm.ravel()

diagnostic_metrics = pd.DataFrame({
    "Metric": [
        "Accuracy", "Sensitivity / pneumonia recall", "Specificity / normal recall",
        "Precision / PPV", "Negative predictive value", "False-positive rate",
        "False-negative rate", "Balanced accuracy"
    ],
    "Value": [
        (tp + tn) / cm.sum(), tp / (tp + fn), tn / (tn + fp),
        tp / (tp + fp), tn / (tn + fn), fp / (fp + tn),
        fn / (fn + tp), 0.5 * (tp / (tp + fn) + tn / (tn + fp))
    ],
    "Interpretation": [
        "All correct predictions / all test images",
        "Ability to detect pneumonia; 1 − miss rate",
        "Ability to correctly reject pneumonia",
        "P(pneumonia | predicted pneumonia)",
        "P(normal | predicted normal)",
        "NORMAL incorrectly flagged as PNEUMONIA",
        "PNEUMONIA incorrectly predicted NORMAL",
        "Mean of sensitivity and specificity"
    ]
})

diagnostic_metrics["Value"] = diagnostic_metrics["Value"].map(lambda x: f"{x:.2%}")
display(diagnostic_metrics)
```

```python
# Counts and row-normalized confusion matrices
cm_normalized = cm / cm.sum(axis=1, keepdims=True)

fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.8))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False,
            xticklabels=class_names, yticklabels=class_names, ax=axes[0])
axes[0].set(title="Confusion matrix — counts", xlabel="Predicted label", ylabel="True label")

sns.heatmap(cm_normalized, annot=True, fmt=".1%", cmap="YlOrRd", vmin=0, vmax=1,
            cbar=False, xticklabels=class_names, yticklabels=class_names, ax=axes[1])
axes[1].set(title="Confusion matrix — row-normalized", xlabel="Predicted label", ylabel="True label")

fig.suptitle("Correct predictions and clinical error types", fontsize=14, fontweight="bold")
fig.tight_layout()
plt.show()
```

```python
# Visual comparison of correct predictions, false alarms, and missed cases
outcome_names = [
    "True negative\nNORMAL→NORMAL", "False positive\nNORMAL→PNEUMONIA",
    "False negative\nPNEUMONIA→NORMAL", "True positive\nPNEUMONIA→PNEUMONIA"
]
outcome_counts = [tn, fp, fn, tp]

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
bars = axes[0].bar(outcome_names, outcome_counts,
                   color=["#4C78A8", "#F58518", "#E45756", "#54A24B"])
axes[0].bar_label(bars, padding=3, fontweight="bold")
axes[0].set(title=f"Test outcomes (n = {cm.sum()})", ylabel="Number of images")
axes[0].tick_params(axis="x", labelrotation=12)
axes[0].grid(axis="y", alpha=0.25)

error_rates = [fp / (tn + fp) * 100, fn / (tp + fn) * 100]
bars = axes[1].bar(["False-alarm rate\n(FPR)", "Miss rate\n(FNR)"], error_rates,
                   color=["#F58518", "#E45756"], width=0.58)
axes[1].bar_label(bars, labels=[f"{value:.2f}%" for value in error_rates],
                  padding=4, fontweight="bold")
axes[1].set(title="Error rates by true class", ylabel="Rate (%)", ylim=(0, 45))
axes[1].grid(axis="y", alpha=0.25)

fig.suptitle("False alarms versus missed pneumonia cases", fontsize=14, fontweight="bold")
fig.tight_layout()
plt.show()
```

### Interpretation of false alarms and missed cases

The model misses only **6 of 390 pneumonia images**, giving a false-negative rate of **1.54%** and pneumonia sensitivity of **98.46%**. This is the main strength of the model for a screening-oriented experiment.

However, it incorrectly flags **91 of 234 normal images**, corresponding to a false-positive rate of **38.89%** and specificity of only **61.11%**. Therefore, approximately two out of every five truly normal test images trigger a false pneumonia alert. The overall accuracy of 84.46% hides this asymmetry; balanced accuracy (79.79%) is the more cautious single-number summary.

The model should therefore be described as **high-sensitivity but low-specificity**. It may reduce missed pneumonia cases in an academic screening demonstration, but its false-alarm burden is too large for autonomous clinical use.

- Bottom row, last image: a true positive.

No false negative appears in that random eight-image sample because false negatives are rare in this run (6/624 test images). The next cells deliberately search the entire test set and display examples from all four outcome groups, including missed pneumonia cases.

```python
# Collect probability scores and image paths for image-level error analysis
def collect_prediction_records(model, loader, dataset, device):
    model.eval()
    probabilities, labels, predictions = [], [], []

    with torch.no_grad():
        for images, batch_labels in loader:
            logits = model(images.to(device))
            batch_probabilities = torch.softmax(logits, dim=1)
            batch_predictions = batch_probabilities.argmax(dim=1)

            probabilities.extend(batch_probabilities.cpu().numpy())
            labels.extend(batch_labels.numpy())
            predictions.extend(batch_predictions.cpu().numpy())

    probabilities = np.asarray(probabilities)
    labels = np.asarray(labels)
    predictions = np.asarray(predictions)
    paths = [sample[0] for sample in dataset.samples]

    records = pd.DataFrame({
        "image_path": paths,
        "true_index": labels,
        "predicted_index": predictions,
        "true_label": [class_names[index] for index in labels],
        "predicted_label": [class_names[index] for index in predictions],
        "p_normal": probabilities[:, 0],
        "p_pneumonia": probabilities[:, 1],
        "confidence": probabilities.max(axis=1),
    })
    records["correct"] = records["true_index"] == records["predicted_index"]
    records["outcome"] = np.select(
        [
            (records.true_index == 0) & (records.predicted_index == 0),
            (records.true_index == 0) & (records.predicted_index == 1),
            (records.true_index == 1) & (records.predicted_index == 0),
            (records.true_index == 1) & (records.predicted_index == 1),
        ],
        ["TN", "FP", "FN", "TP"],
        default="Unknown"
    )
    return records

prediction_records = collect_prediction_records(model, test_loader, test_dataset, device)
display(prediction_records["outcome"].value_counts().reindex(["TN", "FP", "FN", "TP"]))
```

```python
# Display representative images from each confusion-matrix category
def denormalize_image(image_tensor):
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    image = image_tensor.permute(1, 2, 0).cpu().numpy()
    return np.clip(image * std + mean, 0, 1)

def show_outcome_gallery(records, dataset, examples_per_group=3):
    groups = ["TN", "FP", "FN", "TP"]
    fig, axes = plt.subplots(len(groups), examples_per_group,
                             figsize=(4 * examples_per_group, 3.7 * len(groups)))
    dataset_paths = [sample[0] for sample in dataset.samples]
    path_to_index = {path: index for index, path in enumerate(dataset_paths)}

    for row, group in enumerate(groups):
        subset = records[records["outcome"] == group].sort_values("confidence", ascending=False)
        subset = subset.head(examples_per_group)
        for column in range(examples_per_group):
            axis = axes[row, column]
            axis.axis("off")
            if column >= len(subset):
                axis.set_title(f"No {group} example")
                continue
            record = subset.iloc[column]
            image_tensor, _ = dataset[path_to_index[record.image_path]]
            axis.imshow(denormalize_image(image_tensor), cmap="gray")
            title_color = "#2E7D32" if group in {"TN", "TP"} else "#C62828"
            axis.set_title(
                f"{group} | True: {record.true_label}\n"
                f"Pred: {record.predicted_label} | Conf: {record.confidence:.1%}",
                color=title_color, fontsize=10
            )

    fig.suptitle("Representative correct predictions and errors", fontsize=15, fontweight="bold")
    fig.tight_layout()
    plt.show()

show_outcome_gallery(prediction_records, test_dataset, examples_per_group=3)
```

```python
# Visualize the most uncertain predictions (closest to a 50% decision boundary)
uncertain_examples = prediction_records.assign(
    uncertainty=(prediction_records["p_pneumonia"] - 0.5).abs()
).nsmallest(8, "uncertainty")

fig, axes = plt.subplots(2, 4, figsize=(14, 7))
dataset_paths = [sample[0] for sample in test_dataset.samples]
path_to_index = {path: index for index, path in enumerate(dataset_paths)}

for axis, (_, record) in zip(axes.ravel(), uncertain_examples.iterrows()):
    image_tensor, _ = test_dataset[path_to_index[record.image_path]]
    axis.imshow(denormalize_image(image_tensor), cmap="gray")
    axis.set_title(
        f"True: {record.true_label} | Pred: {record.predicted_label}\n"
        f"P(PNEUMONIA) = {record.p_pneumonia:.1%}",
        color="#2E7D32" if record.correct else "#C62828", fontsize=9
    )
    axis.axis("off")

fig.suptitle("Most uncertain test predictions", fontsize=15, fontweight="bold")
fig.tight_layout()
plt.show()
```

### Probability-based evaluation: ROC, precision–recall, and threshold trade-offs

Accuracy, sensitivity and specificity describe the default 0.50 decision threshold. The following analysis evaluates the model across all thresholds. ROC-AUC measures ranking ability, while average precision and the precision–recall curve are especially useful when class frequencies are unequal. The threshold table makes the sensitivity–specificity trade-off explicit instead of treating 0.50 as automatically optimal.

```python
# ROC curve, precision–recall curve, and probability-score metrics
from sklearn.metrics import (
    roc_curve, roc_auc_score, precision_recall_curve,
    average_precision_score, auc
)

y_true = prediction_records["true_index"].to_numpy()
y_score = prediction_records["p_pneumonia"].to_numpy()

fpr_curve, tpr_curve, _ = roc_curve(y_true, y_score)
precision_curve, recall_curve, _ = precision_recall_curve(y_true, y_score)
roc_auc = roc_auc_score(y_true, y_score)
average_precision = average_precision_score(y_true, y_score)

fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.8))
axes[0].plot(fpr_curve, tpr_curve, label=f"ROC-AUC = {roc_auc:.3f}")
axes[0].plot([0, 1], [0, 1], "--", color="gray", label="Random classifier")
axes[0].set(title="ROC curve", xlabel="False-positive rate", ylabel="True-positive rate")
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(recall_curve, precision_curve, label=f"Average precision = {average_precision:.3f}")
axes[1].axhline(y_true.mean(), linestyle="--", color="gray", label=f"Prevalence = {y_true.mean():.3f}")
axes[1].set(title="Precision–recall curve", xlabel="Recall", ylabel="Precision")
axes[1].legend()
axes[1].grid(alpha=0.3)

fig.suptitle("Threshold-independent discrimination", fontsize=14, fontweight="bold")
fig.tight_layout()
plt.show()
```

```python
# Threshold sensitivity analysis
threshold_rows = []
for threshold in np.arange(0.10, 0.91, 0.05):
    threshold_predictions = (y_score >= threshold).astype(int)
    threshold_cm = confusion_matrix(y_true, threshold_predictions, labels=[0, 1])
    threshold_tn, threshold_fp, threshold_fn, threshold_tp = threshold_cm.ravel()
    sensitivity = threshold_tp / (threshold_tp + threshold_fn)
    specificity = threshold_tn / (threshold_tn + threshold_fp)
    precision = threshold_tp / max(threshold_tp + threshold_fp, 1)
    threshold_rows.append({
        "threshold": threshold,
        "sensitivity": sensitivity,
        "specificity": specificity,
        "precision": precision,
        "balanced_accuracy": 0.5 * (sensitivity + specificity),
        "false_positives": threshold_fp,
        "false_negatives": threshold_fn,
    })

threshold_table = pd.DataFrame(threshold_rows)
best_threshold_row = threshold_table.loc[threshold_table["balanced_accuracy"].idxmax()]
display(threshold_table.style.format({
    "threshold": "{:.2f}", "sensitivity": "{:.2%}", "specificity": "{:.2%}",
    "precision": "{:.2%}", "balanced_accuracy": "{:.2%}"
}).highlight_max(subset=["balanced_accuracy"], color="#C8E6C9"))

plt.figure(figsize=(9, 5))
for metric in ["sensitivity", "specificity", "precision", "balanced_accuracy"]:
    plt.plot(threshold_table["threshold"], threshold_table[metric], marker="o", label=metric)
plt.axvline(0.50, color="black", linestyle="--", alpha=0.6, label="Default threshold")
plt.axvline(best_threshold_row.threshold, color="#2E7D32", linestyle=":", linewidth=2,
            label=f"Best balanced threshold = {best_threshold_row.threshold:.2f}")
plt.xlabel("Pneumonia probability threshold")
plt.ylabel("Metric value")
plt.ylim(0, 1.03)
plt.title("Sensitivity–specificity trade-off")
plt.grid(alpha=0.3)
plt.legend(loc="best")
plt.tight_layout()
plt.show()
```

### Evaluation caveats

1. **Threshold tuning must use validation data, not the test set.** The threshold table is exploratory; selecting a production threshold from test performance would leak test information.
2. **The validation set contains only 16 images.** This makes checkpoint and threshold selection unstable. A larger patient-level validation split is the highest-priority improvement.
3. **Probabilities are not automatically calibrated.** A high softmax confidence is not a clinical probability. Reliability diagrams, expected calibration error, and temperature scaling should be added only after obtaining a sufficiently large validation set.
4. **Image examples are qualitative evidence.** They help inspect failure modes but do not replace aggregate metrics or expert radiological review.
5. **No clinical claim is made.** The experiment is appropriate for coursework, not diagnosis or autonomous decision-making.

## 8. Augmentation assessment

The notebook demonstrates an augmentation pipeline but does not contain a matched no-augmentation training run. Therefore, the report can conclude that augmentation was **used**, but cannot claim that it improved accuracy.

A defensible augmentation experiment must keep the split, architecture, optimizer, learning rate, seed, batch size and epochs fixed, then compare:

1. E1: no random augmentation.
2. E2: horizontal flip + mild rotation.

The same test metrics should be reported for both models. Only then can the group state whether augmentation improved generalization.


## 9. Error analysis and limitations

### Evidence-based error summary

- **False alarms dominate:** 91 normal X-rays were classified as pneumonia (FPR 38.89%).
- **Missed pneumonia is uncommon but not zero:** 6 pneumonia X-rays were classified as normal (FNR 1.54%).
- The sensitivity–specificity imbalance means accuracy alone is insufficient; balanced accuracy and both error rates must be reported.
- The 16-image validation set makes checkpoint selection unstable.
- The training set is imbalanced toward pneumonia.
- The folder split does not prove patient-level independence, so leakage cannot be excluded.
- The 25.78-million-parameter fully connected head increases overfitting risk.

### Recommended improvements

1. Rebuild the train/validation/test split at patient level and enlarge validation.
2. Compare augmentation against an otherwise identical no-augmentation baseline.
3. Replace flattening with Global Average Pooling or use a regularized pretrained backbone.
4. Compare class weighting, balanced sampling, and focal loss using macro F1 and balanced accuracy.
5. Tune the classification threshold only on validation data and lock it before final testing.
6. Add calibration analysis and confidence intervals after increasing validation size.
7. Validate on an independent hospital dataset and obtain radiologist review before discussing clinical use.

## 10. Conclusion

The CNN achieved **84.46% test accuracy**, **98.46% pneumonia sensitivity**, and **61.11% specificity**. It correctly detected 384 of 390 pneumonia images and missed 6, but incorrectly flagged 91 of 234 normal images. The resulting **1.54% miss rate** is low, whereas the **38.89% false-alarm rate** is high. Balanced accuracy is **79.79%**, which better reflects the unequal class performance than raw accuracy alone.

The correct conclusion is therefore not simply that the model is “84.46% accurate.” It is a **high-sensitivity, low-specificity classifier** that prioritizes pneumonia detection at the cost of many false alarms. The coursework demonstrates a complete deep-learning workflow—data inspection, augmentation, CNN training, checkpoint selection, class-wise evaluation, confusion-matrix analysis, and qualitative inspection of correct and incorrect predictions—but does not support clinical deployment.
